```python
# IMPORT PACKAGES

from pathlib import Path
import shutil

import rasterio
from rasterio.shutil import copy as rio_copy

from tqdm import tqdm
```

```python
# ROOT DIRECTORY

ROOT_DIR = Path("..")

# INPUT

INPUT_DIR = ROOT_DIR / "dataset_clipped"

# OUTPUT

OUTPUT_DIR = ROOT_DIR / "dataset_aligned"

# CREATE OUTPUT

OUTPUT_DIR.mkdir(exist_ok=True)

print("Paths ready.")
```

```python
# FIND YEARS

year_folders = sorted([
    f for f in INPUT_DIR.iterdir()
    if f.is_dir()
])

print(year_folders)
```

```python
def align_dataset():

    for year_folder in year_folders:

        year_name = year_folder.name

        print(f"\nPROCESSING YEAR: {year_name}")

        # CREATE OUTPUT YEAR
        output_year = OUTPUT_DIR / year_name
        output_year.mkdir(exist_ok=True)

        # FIND FLIGHTS
        flight_folders = sorted([
            f for f in year_folder.iterdir()
            if f.is_dir()
        ])

        for flight_folder in tqdm(flight_folders):

            flight_name = flight_folder.name

            print(f"\nPROCESSING FLIGHT: {flight_name}")

            # DATE STRING
            date_string = flight_name.replace("flight_", "")

            # OUTPUT FLIGHT
            output_flight = output_year / flight_name
            output_flight.mkdir(exist_ok=True)

            # -----------------------------------------
            # CHECK IF ALREADY PROCESSED
            # -----------------------------------------

            expected_output = (
                output_flight /
                f"{date_string}_nir_reflectance.tif"
            )

            if expected_output.exists():

                print("Flight already processed. Skipping.")

                continue

            # FIND TIFFS
            tif_files = list(flight_folder.glob("*.tif"))

            for tif_path in tif_files:

                filename = tif_path.name.lower()

                # -----------------------------------------
                # RGB VISUALIZATION
                # -----------------------------------------

                if "group1" in filename:

                    output_name = (
                        f"{date_string}_rgb_visualization.tif"
                    )

                    shutil.copy2(
                        tif_path,
                        output_flight / output_name
                    )

                    print("Copied RGB visualization")

                # -----------------------------------------
                # GREEN REFLECTANCE
                # -----------------------------------------

                elif (
                    "green" in filename
                    and "group1" not in filename
                ):

                    output_name = (
                        f"{date_string}_green_reflectance.tif"
                    )

                    shutil.copy2(
                        tif_path,
                        output_flight / output_name
                    )

                    print("Copied GREEN reflectance")

                # -----------------------------------------
                # RED REFLECTANCE
                # -----------------------------------------

                elif (
                    "red" in filename
                    and "red edge" not in filename
                    and "red_edge" not in filename
                    and "group1" not in filename
                ):

                    output_name = (
                        f"{date_string}_red_reflectance.tif"
                    )

                    shutil.copy2(
                        tif_path,
                        output_flight / output_name
                    )

                    print("Copied RED reflectance")

                # -----------------------------------------
                # NIR REFLECTANCE
                # -----------------------------------------

                elif "nir" in filename:

                    output_name = (
                        f"{date_string}_nir_reflectance.tif"
                    )

                    shutil.copy2(
                        tif_path,
                        output_flight / output_name
                    )

                    print("Copied NIR reflectance")

                # -----------------------------------------
                # RED EDGE REFLECTANCE
                # -----------------------------------------

                elif (
                    "red edge" in filename
                    or "red_edge" in filename
                    or "rededge" in filename
                ):

                    output_name = (
                        f"{date_string}_rededge_reflectance.tif"
                    )

                    shutil.copy2(
                        tif_path,
                        output_flight / output_name
                    )

                    print("Copied RED EDGE reflectance")

                # -----------------------------------------
                # THERMAL
                # -----------------------------------------

                elif "thermal" in filename:

                    output_name = (
                        f"{date_string}_thermal.tif"
                    )

                    shutil.copy2(
                        tif_path,
                        output_flight / output_name
                    )

                    print("Copied THERMAL")

    print("\nDATASET ALIGNMENT COMPLETED.")
```

```python
align_dataset()
```
