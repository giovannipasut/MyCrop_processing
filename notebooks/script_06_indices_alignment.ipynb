```python
# IMPORT PACKAGES

from pathlib import Path

import numpy as np
import rasterio

from tqdm import tqdm
```

```python
# ROOT DIRECTORY

ROOT_DIR = Path("..")

# INPUT
INPUT_DIR = ROOT_DIR / "indices"

# OUTPUT
OUTPUT_DIR = ROOT_DIR / "indices_aligned"

OUTPUT_DIR.mkdir(exist_ok=True)

print("Paths ready.")
```

```python
# FIND YEARS

year_folders = sorted([
    f for f in INPUT_DIR.iterdir()
    if f.is_dir()
])

print(year_folders)
```

```python
def clean_index(data):

    data = data.astype("float32")

    # REMOVE inf
    data[np.isinf(data)] = np.nan

    # REMOVE invalid values
    data[data > 1] = np.nan
    data[data < -1] = np.nan

    # REMOVE zero values
    data[data == 0] = np.nan

    return data
```

```python
def align_indices():

    for year_folder in year_folders:

        year_name = year_folder.name

        print(f"\nPROCESSING YEAR: {year_name}")

        # CREATE OUTPUT YEAR
        output_year = OUTPUT_DIR / year_name
        output_year.mkdir(exist_ok=True)

        # FIND FLIGHTS
        flight_folders = sorted([
            f for f in year_folder.iterdir()
            if f.is_dir()
        ])

        for flight_folder in tqdm(flight_folders):

            flight_name = flight_folder.name

            print(f"\nPROCESSING FLIGHT: {flight_name}")

            # DATE STRING
            date_string = flight_name.replace("flight_", "")

            # CREATE OUTPUT FLIGHT
            output_flight = output_year / flight_name
            output_flight.mkdir(exist_ok=True)

            # -----------------------------------------
            # CHECK IF ALREADY PROCESSED
            # -----------------------------------------

            expected_output = (
                output_flight /
                f"{date_string}_NDVI.tif"
            )

            if expected_output.exists():

                print("Flight already processed. Skipping.")

                continue

            # FIND TIFF FILES
            tif_files = list(flight_folder.glob("*.tif"))

            for tif_path in tif_files:

                filename = tif_path.name.lower()

                print(f"Processing: {filename}")

                # -----------------------------------------
                # GNDVI
                # -----------------------------------------

                if "gndvi" in filename:

                    output_name = (
                        f"{date_string}_GNDVI.tif"
                    )

                    with rasterio.open(tif_path) as src:

                        data = src.read(1)

                        profile = src.profile

                    cleaned_data = clean_index(data)

                    profile.update(
                        dtype=rasterio.float32,
                        nodata=np.nan,
                        compress="lzw"
                    )

                    output_path = (
                        output_flight / output_name
                    )

                    with rasterio.open(
                        output_path,
                        "w",
                        **profile
                    ) as dst:

                        dst.write(
                            cleaned_data.astype(
                                rasterio.float32
                            ),
                            1
                        )

                    print("Saved GNDVI")

                # -----------------------------------------
                # NDVI
                # -----------------------------------------

                elif (
                    "ndvi" in filename
                    and "gndvi" not in filename
                ):

                    output_name = (
                        f"{date_string}_NDVI.tif"
                    )

                    with rasterio.open(tif_path) as src:

                        data = src.read(1)

                        profile = src.profile

                    cleaned_data = clean_index(data)

                    profile.update(
                        dtype=rasterio.float32,
                        nodata=np.nan,
                        compress="lzw"
                    )

                    output_path = (
                        output_flight / output_name
                    )

                    with rasterio.open(
                        output_path,
                        "w",
                        **profile
                    ) as dst:

                        dst.write(
                            cleaned_data.astype(
                                rasterio.float32
                            ),
                            1
                        )

                    print("Saved NDVI")

                # -----------------------------------------
                # NDRE
                # -----------------------------------------

                elif "ndre" in filename:

                    output_name = (
                        f"{date_string}_NDRE.tif"
                    )

                    with rasterio.open(tif_path) as src:

                        data = src.read(1)

                        profile = src.profile

                    cleaned_data = clean_index(data)

                    profile.update(
                        dtype=rasterio.float32,
                        nodata=np.nan,
                        compress="lzw"
                    )

                    output_path = (
                        output_flight / output_name
                    )

                    with rasterio.open(
                        output_path,
                        "w",
                        **profile
                    ) as dst:

                        dst.write(
                            cleaned_data.astype(
                                rasterio.float32
                            ),
                            1
                        )

                    print("Saved NDRE")

                # -----------------------------------------
                # CANOPY TEMPERATURE
                # -----------------------------------------

                elif "canopy_temperature" in filename:

                    output_name = (
                        f"{date_string}_canopy_temperature.tif"
                    )

                    with rasterio.open(tif_path) as src:

                        data = src.read(1)

                        profile = src.profile

                    profile.update(
                        dtype=rasterio.float32,
                        compress="lzw"
                    )

                    output_path = (
                        output_flight / output_name
                    )

                    with rasterio.open(
                        output_path,
                        "w",
                        **profile
                    ) as dst:

                        dst.write(
                            data.astype(
                                rasterio.float32
                            ),
                            1
                        )

                    print("Saved canopy temperature")

                # -----------------------------------------
                # UNKNOWN FILE
                # -----------------------------------------

                else:

                    print("Skipping unknown file")
                    continue

    print("\nINDICES ALIGNMENT COMPLETED.")
```

```python
align_indices()
```
