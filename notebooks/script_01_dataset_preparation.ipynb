```python
#IMPORT OF THE VARIOUS PACKAGES
import os
import shutil
from pathlib import Path
from tqdm import tqdm
```

```python
# Root folder
ROOT_DIR = Path("..")

# Original data folders
FLIGHTS_2025 = ROOT_DIR / "data" / "2025_flights"
FLIGHTS_2026 = ROOT_DIR / "data" / "2026_flights"

# Output dataset folder
DATASET_DIR = ROOT_DIR / "dataset"

# Create dataset folder if not exists
DATASET_DIR.mkdir(exist_ok=True)

print("Dataset folder ready.")
```

```python
def process_year_folder(year_folder, year_name):

    # Create output folder for the year
    output_year_folder = DATASET_DIR / year_name
    output_year_folder.mkdir(exist_ok=True)

    # Find all flight folders
    flight_folders = sorted([
        f for f in year_folder.iterdir()
        if f.is_dir() and f.name.startswith("flight_")
    ])

    print(f"\nFound {len(flight_folders)} flights in {year_name}")

    # Loop through flights
    for flight_folder in tqdm(flight_folders):

        flight_name = flight_folder.name

        # Output folder for this flight
        output_flight_folder = output_year_folder / flight_name

        # -------------------------------------------------
        # CHECK IF FLIGHT ALREADY EXISTS
        # -------------------------------------------------

        if output_flight_folder.exists():

            existing_tifs = list(output_flight_folder.glob("*.tif"))

            if len(existing_tifs) > 0:
                print(f"\nSKIPPING {flight_name} -> already processed")
                continue

        # Create output flight folder
        output_flight_folder.mkdir(exist_ok=True)

        print(f"\nPROCESSING {flight_name}")

        # Find subfolders inside flight folder
        subfolders = [
            folder for folder in flight_folder.iterdir()
            if folder.is_dir()
        ]

        # Loop through subfolders
        for subfolder in subfolders:

            # Path to reflectance folder
            reflectance_path = subfolder / "4_index" / "reflectance"

            # Check if reflectance exists
            if reflectance_path.exists():

                # Find all tif files
                tif_files = list(reflectance_path.glob("*.tif"))

                print(f"  Found {len(tif_files)} tif files")

                # Loop through tif files
                for tif_file in tif_files:

                    filename = tif_file.name

                    # Destination path
                    destination = output_flight_folder / filename

                    # Avoid duplicates
                    if destination.exists():
                        print(f"    Already exists: {filename}")
                        continue

                    # Copy tif
                    shutil.copy2(tif_file, destination)

                    print(f"    Copied: {filename}")

            else:
                print(f"Reflectance folder NOT FOUND -> {reflectance_path}")

    print(f"\nFinished processing {year_name}")
```

```python
process_year_folder(FLIGHTS_2025, "2025")

process_year_folder(FLIGHTS_2026, "2026")

print("\nDataset preparation completed.")
```
